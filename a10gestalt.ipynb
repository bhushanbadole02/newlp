```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.animation import FuncAnimation
```

```python
# Mapping textures (marker styles)
texture_map = {'A': 'o', 'B': '^', 'C': 's'}
color_map = {'A': 'blue', 'B': 'green', 'C': 'red'}
```

```python
# === 1. Scatterplot with Gestalt Laws + Transparency ===
# Define example data for x, y, size, and category
np.random.seed(42)
num_samples = 100
x = np.random.rand(num_samples) * 10
y = np.random.rand(num_samples) * 10
size = np.random.randint(10, 200, num_samples)
category = np.random.choice(['A', 'B', 'C'], num_samples)

plt.figure(figsize=(8,6))
for cat in np.unique(category):
    mask = category == cat
    plt.scatter(x[mask], y[mask],
                alpha=0.5,                       # Transparency for overlapping
                s=size[mask] / 2,               # Mapping size dimension
                c=color_map[cat],
                marker=texture_map[cat],        # Texture patterns
                label=f"Category {cat}")

plt.title("Scatterplot Using Gestalt Laws, Texture, and Transparency")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# === 2. Multidimensional Pattern Visualization ===
# Define example data for x, y, and size
np.random.seed(42)
num_samples = 50
x = np.random.rand(num_samples) * 10
y = np.random.rand(num_samples) * 10
size = np.random.randint(10, 200, num_samples)

plt.figure(figsize=(8,6))
scatter = plt.scatter(x, y,
                      c=size, cmap='viridis',
                      alpha=0.6,
                      s=(size/4))

plt.colorbar(scatter, label="Size Dimension")
plt.title("Multidimensional Discrete Data Pattern Visualization")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
# === 3. Patterns in Motion (Animation) ===
fig, ax = plt.subplots(figsize=(8,6))
ax.set_xlim(-4, 4)
ax.set_ylim(-4, 4)
```

```python
# Initial positions for animated group
num_points = 30
x_anim = np.random.randn(num_points)
y_anim = np.random.randn(num_points)

points = ax.scatter(x_anim, y_anim, c='purple', alpha=0.7)

def update(frame):
    global x_anim, y_anim
    x_anim += np.random.uniform(-0.1, 0.1, num_points)
    y_anim += np.random.uniform(-0.1, 0.1, num_points)
    points.set_offsets(np.c_[x_anim, y_anim])
    return points,

ani = FuncAnimation(fig, update, frames=100, interval=100)
plt.title("Patterns in Motion (Gestalt: Common Fate)")
plt.show()
```
