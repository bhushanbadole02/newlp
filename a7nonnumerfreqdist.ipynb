```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Sample dataset with categorical variables
data = {
    'Department': ['HR', 'Finance', 'IT', 'IT', 'Finance', 'HR', 'IT', 'Marketing', 'Marketing', 'HR'],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Female', 'Male', 'Male', 'Female', 'Female', 'Male']
}

df = pd.DataFrame(data)

print("Dataset:")
print(df)
```

```python
# -------------------------------------------
# a) FREQUENCY DISTRIBUTION
# -------------------------------------------

dept_freq = df['Department'].value_counts()
gender_freq = df['Gender'].value_counts()

print("\nFrequency Distribution (Department):")
print(dept_freq)

print("\nFrequency Distribution (Gender):")
print(gender_freq)
```

```python
# Bar plot for frequency distribution
plt.figure(figsize=(6,4))
sns.countplot(x='Department', data=df)
plt.title("Frequency Distribution of Departments")
plt.show()

plt.figure(figsize=(6,4))
sns.countplot(x='Gender', data=df)
plt.title("Frequency Distribution of Gender")
plt.show()
```

```python
# -------------------------------------------
# b) PROPORTION OF EACH CATEGORY
# -------------------------------------------

plt.figure(figsize=(6,6))
dept_freq.plot(kind='pie', autopct='%1.1f%%')
plt.title("Proportion of Departments")
plt.ylabel("")   # Hide y label
plt.show()

plt.figure(figsize=(6,6))
gender_freq.plot(kind='pie', autopct='%1.1f%%')
plt.title("Proportion of Gender")
plt.ylabel("")
plt.show()
```

```python
# -------------------------------------------
# c) CATEGORY SPLIT BY ANOTHER CATEGORY
# -------------------------------------------

plt.figure(figsize=(7,5))
sns.countplot(x='Department', hue='Gender', data=df)
plt.title("Department Distribution Split by Gender")
plt.show()
```
