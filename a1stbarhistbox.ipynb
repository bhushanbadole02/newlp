```python
# Importing required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load dataset (Iris dataset)
from sklearn.datasets import load_iris
data = load_iris()
df = pd.DataFrame(data.data, columns=data.feature_names)
```

```python
# -------------------------
# A) Compute Summary Statistics
# -------------------------

summary_stats = pd.DataFrame({
    'Minimum': df.min(),
    'Maximum': df.max(),
    'Mean': df.mean(),
    'Range': df.max() - df.min(),
    'Standard Deviation': df.std(),
    'Variance': df.var(),
    '25th Percentile': df.quantile(0.25),
    '50th Percentile': df.quantile(0.50),
    '75th Percentile': df.quantile(0.75)
})

print("Summary Statistics:\n")
print(summary_stats)
```

```python
# Plotting bar graph of mean values
plt.figure(figsize=(10, 6))
plt.bar(df.columns, df.mean())
plt.title("Mean Values of Each Feature")
plt.xlabel("Features")
plt.ylabel("Mean")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# -------------------------
# B) Histogram for Each Feature
# -------------------------

df.hist(figsize=(10, 8), bins=10)
plt.suptitle("Histograms of Dataset Features")
plt.tight_layout()
plt.show()
```

```python
# -------------------------
# C) Combined Boxplot
# -------------------------

plt.figure(figsize=(10, 6))
sns.boxplot(data=df, orient='v')
plt.title("Combined Boxplot of All Features")
plt.xlabel("Features")
plt.ylabel("Values")
plt.xticks(rotation=45)
plt.show()
```
