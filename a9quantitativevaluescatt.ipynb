```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Create a sample dataset
np.random.seed(42)
data = pd.DataFrame({
    "Height": np.random.normal(165, 10, 150),
    "Weight": np.random.normal(65, 12, 150),
    "Region": np.random.choice(["North", "South", "East", "West"], 150)
})
```

```python
# Order categories manually
region_order = ["North", "South", "East", "West"]
data["Region"] = pd.Categorical(data["Region"], categories=region_order, ordered=True)
```

```python
# ---- Basic Scatterplot ----
plt.figure(figsize=(7,5))
plt.scatter(data["Height"], data["Weight"])
plt.title("Basic Scatterplot of Height vs Weight")
plt.xlabel("Height (cm)")
plt.ylabel("Weight (kg)")
plt.show()
```

```python
# ---- Scatterplot with Categorical Separation using Color ----
plt.figure(figsize=(7,5))
sns.scatterplot(data=data, x="Height", y="Weight", hue="Region", palette="Set2")
plt.title("Scatterplot with Categorical Separation (Coloring by Region)")
plt.show()
```

```python
# ---- Categorical Alignment Using FacetGrid (Subplots) ----
g = sns.FacetGrid(data, col="Region", col_order=region_order)
g.map_dataframe(sns.scatterplot, x="Height", y="Weight")
g.add_legend()
plt.suptitle("Scatterplot Alignment by Region (Faceted View)", y=1.03)
plt.show()
```

```python
# ---- Categorical Alignment Using Strip Plot (Jittered) ----
plt.figure(figsize=(8,5))
sns.stripplot(data=data, x="Region", y="Weight", jitter=True, order=region_order)
plt.title("Strip Plot Showing Weight Distribution by Region")
plt.show()
```
