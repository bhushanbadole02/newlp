```python
import pandas as pd
import numpy as np
from scipy.stats import skew, kurtosis
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Step 1: Load Dataset (Sample dataset - Iris)
from sklearn.datasets import load_iris
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
```

```python
# Step 2: Display first few rows
print("First 5 rows of dataset:")
print(df.head())
```

```python
# Step 3: Descriptive Statistics
print("\nDescriptive Statistics:")
print(df.describe())
```

```python
# Step 4: Skewness of each column
print("\nSkewness of each column:")
print(df.apply(skew))
```

```python
# Step 5: Kurtosis of each column
print("\nKurtosis of each column:")
print(df.apply(kurtosis))
```

```python
# Step 6: Correlation Matrix
print("\nCorrelation Matrix:")
corr_matrix = df.corr()
print(corr_matrix)
```

```python
# Step 7: Heatmap Visualization
plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```
