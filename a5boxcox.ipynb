```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import boxcox, skew, probplot
```

```python
# Step 1: Create a sample positively skewed dataset
np.random.seed(42)
data = np.random.exponential(scale=2, size=500)
df = pd.DataFrame({'Value': data})
```

```python
# Plot before transformation
plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
sns.histplot(df['Value'], kde=True)
plt.title("Before Box-Cox Transformation")

plt.subplot(1, 2, 2)
probplot(df['Value'], dist="norm", plot=plt)
plt.title("Q-Q Plot Before Transformation")

plt.show()

print("Skewness before:", skew(df['Value']))
```

```python
# Step 2: Apply Box-Cox transformation
boxcox_transformed, lambda_optimal = boxcox(df['Value'])

df['Transformed'] = boxcox_transformed

print("Optimal λ:", lambda_optimal)
```

```python
# Plot after transformation
plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
sns.histplot(df['Transformed'], kde=True)
plt.title("After Box-Cox Transformation")

plt.subplot(1, 2, 2)
probplot(df['Transformed'], dist="norm", plot=plt)
plt.title("Q-Q Plot After Transformation")

plt.show()

print("Skewness after:", skew(df['Transformed']))
```
