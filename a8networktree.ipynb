```python
import networkx as nx
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
# -------------------------------
# Part 1: Create a sample network
# -------------------------------

G = nx.Graph()

edges = [
    ('A','B'), ('A','C'), ('B','D'), ('C','D'),
    ('C','E'), ('E','F'), ('F','G'), ('B','G')
]
G.add_edges_from(edges)

plt.figure(figsize=(7,6))
pos = nx.spring_layout(G, seed=42)
nx.draw(G, pos, with_labels=True, node_color='lightblue', edge_color='gray', node_size=700)
plt.title("Network Graph Using Link Marks")
plt.show()
```

```python
# -------------------------------
# Part 2: Adjacency Matrix View
# -------------------------------

adj_matrix = nx.to_numpy_array(G)
labels = list(G.nodes())
df_matrix = pd.DataFrame(adj_matrix, index=labels, columns=labels)

plt.figure(figsize=(8,6))
sns.heatmap(df_matrix, cmap="Blues", annot=True, linewidths=.5)
plt.title("Adjacency Matrix Heatmap")
plt.show()
```

```python
# -------------------------------
# Part 3: Create and Visualize a Tree
# -------------------------------

T = nx.balanced_tree(r=2, h=3)  # Binary tree of height 3

plt.figure(figsize=(7,6))
pos_tree = nx.spring_layout(T, seed=42)
nx.draw(T, pos_tree, with_labels=True, node_color='lightgreen', edge_color='brown')
plt.title("Tree Structure Visualization")
plt.show()
```

```python
# -------------------------------
# Part 4: Findings / Analysis
# -------------------------------

print("Nodes in the Network Graph:", G.nodes())
print("Edges in the Network Graph:", G.edges())
print("\nAdjacency Matrix:\n", df_matrix)
```
